### 2.1 Khám phá Tọa độ không gian 3D & Phân bố Ô kệ (Storage Location)

```python
# Thống kê mô tả tọa độ không gian 3D
print("=== THỐNG KÊ TỌA ĐỘ 3D (x, y, z) ===")
display(df_location[['x', 'y', 'z']].describe())

# Trích xuất Zone, Aisle, Slot từ originalLocation (Format: Zone-Aisle-Slot)
loc_parsed = df_location['originalLocation'].str.extract(r'([A-Za-z]+)-(\d+)-(\d+)')
loc_parsed.columns = ['Zone', 'Aisle', 'Slot']
df_location_detail = pd.concat([df_location, loc_parsed], axis=1)

print("\n=== PHÂN BỐ VỊ TRÍ THEO ZONE (KHU VỰC) ===")
print(df_location_detail['Zone'].value_counts())

print("\n=== PHÂN BỐ THEO TẦNG KỆ (TRỤC Z) ===")
print(df_location['z'].value_counts().sort_index())
```

```python
import numpy as np

# Tính khoảng cách Euclidean từ gốc tọa độ (0, 0, 0)
# Khoảng cách này phản ánh cự ly từ cửa vào/xuất kho đến vị trí ô kệ
df_location['dist_from_origin'] = np.sqrt(
    df_location['x']**2 + df_location['y']**2 + (df_location['z'] * 10)**2
).round(2)

print("=== THỐNG KÊ KHOẢNG CÁCH TỪ GỐC (dist_from_origin) ===")
print(df_location['dist_from_origin'].describe())

print("\nTop 5 vị trí gần cửa xuất nhất:")
display(df_location.sort_values('dist_from_origin').head(5))

print("\nTop 5 vị trí xa cửa xuất nhất:")
display(df_location.sort_values('dist_from_origin', ascending=False).head(5))
```

## 3. Khám Phá Dataset Customer_Order.csv (Đơn Hàng Khách Hàng)

```python
# Đọc Customer_Order.csv
df_order = pd.read_csv(F_ORDER, sep=';', encoding='utf-8-sig')

print("Shape:", df_order.shape)
print("\n5 dòng đầu:")
display(df_order.head())

print("\nThông tin dữ liệu:")
df_order.info()
```

```python
print("=== KIỂM TRA DUPLICATE & MISSING VALUES ===")
print("Số dòng duplicate toàn bộ:", df_order.duplicated().sum())

print("\nSố lượng giá trị Missing (Null) theo từng cột:")
null_series = df_order.isnull().sum()
print(null_series)

if null_series['Size (US)'] > 0:
    print(f"\nPhát hiện {null_series['Size (US)']} dòng thiếu Size (US). Mẫu:")
    display(df_order[df_order['Size (US)'].isnull()].head(3))
```

```python
print("=== THỐNG KÊ UNIQUE CỦA CUSTOMER_ORDER ===")
print("Số khách hàng (codCustomer)   :", df_order['codCustomer'].nunique())
print("Số đơn hàng (orderNumber)     :", df_order['orderNumber'].nunique())
print("Số sản phẩm (Reference)       :", df_order['Reference'].nunique())
print("Số đợt lấy hàng (waveNumber)  :", df_order['waveNumber'].nunique())
print("Số nhân viên (operator)       :", df_order['operator'].nunique())

print("\n=== THỐNG KÊ SỐ LƯỢNG ĐẶT (quantity units) & SIZE (US) ===")
display(df_order[['quantity (units)', 'Size (US)']].describe())
```

```python
# Phân tích chuỗi thời gian đơn hàng (creationDate)
df_order['creationDate_dt'] = pd.to_datetime(df_order['creationDate'], format='%d/%m/%Y %H:%M')

print("=== KHOẢNG THỜI GIAN ĐƠN HÀNG ===")
print("Đơn hàng sớm nhất:", df_order['creationDate_dt'].min())
print("Đơn hàng muộn nhất:", df_order['creationDate_dt'].max())

# Đếm số lượng đơn hàng theo tháng
df_order['order_month'] = df_order['creationDate_dt'].dt.to_period('M')
orders_per_month = df_order.groupby('order_month')['orderNumber'].nunique()

print("\n=== SỐ LƯỢNG ĐƠN HÀNG THEO THÁNG ===")
print(orders_per_month)
```

```python
print("=== TOP 10 KHÁCH HÀNG ĐẶT NHIỀU ĐƠN NHẤT ===")
top_customers = df_order.groupby('codCustomer').agg(
    total_orders=('orderNumber', 'nunique'),
    total_items=('quantity (units)', 'sum')
).sort_values('total_items', ascending=False).head(10)
display(top_customers)

print("\n=== TOP 10 SẢN PHẨM ĐƯỢC ĐẶT NHIỀU NHẤT ===")
top_products = df_order.groupby('Reference').agg(
    order_count=('orderNumber', 'count'),
    total_quantity=('quantity (units)', 'sum')
).sort_values('total_quantity', ascending=False).head(10)
display(top_products)
```

## 6. Khám Phá Nhóm Dataset Chiến Lược Lưu Trữ (Storage Strategies)
Bao gồm 4 chiến lược: **Random**, **Class-Based**, **Dedicated**, **Hybrid**.

```python
# Đọc 4 file chiến lược lưu trữ
df_random     = pd.read_csv(F_RANDOM)
df_classbased = pd.read_csv(F_CLASSBASED, sep=';', encoding='utf-8-sig')
df_dedicated  = pd.read_csv(F_DEDICATED,  sep=';', encoding='utf-8-sig')
df_hybrid     = pd.read_csv(F_HYBRID,     sep=';', encoding='utf-8-sig')

storage_summary = pd.DataFrame([
    {"Strategy": "Random",     "File": "Random_Storage.csv",     "Rows": df_random.shape[0],     "Cols": df_random.shape[1], "Key Col": "originalLocation"},
    {"Strategy": "Class-Based","File": "Class_Based_Storage.csv","Rows": df_classbased.shape[0], "Cols": df_classbased.shape[1], "Key Col": "Location + ABCCOD"},
    {"Strategy": "Dedicated",  "File": "Dedicated_Storage.csv",  "Rows": df_dedicated.shape[0],  "Cols": df_dedicated.shape[1], "Key Col": "Location + XYZCOD"},
    {"Strategy": "Hybrid",     "File": "Hybrid_Storage.csv",     "Rows": df_hybrid.shape[0],     "Cols": df_hybrid.shape[1], "Key Col": "Location + XYZCOD"},
])

print("=== SO SÁNH TỔNG QUAN 4 CHIẾN LƯỢC LƯU TRỮ ===")
display(storage_summary)
```

```python
print("=== 1. RANDOM STORAGE (Không phân nhóm) ===")
print("5 dòng đầu:")
display(df_random.head(3))
print("Ví dụ nội dung ô slot (col_1):", df_random['col_1'].head(5).tolist())

print("\n=== 2. CLASS-BASED STORAGE (Phân nhóm ABC) ===")
print("Phân bố số vị trí ô kệ theo nhóm ABCCOD:")
print(df_classbased['ABCCOD'].value_counts())
print("\nTỷ lệ (%):")
print((df_classbased['ABCCOD'].value_counts(normalize=True) * 100).round(2))
```

```python
print("=== 3. DEDICATED STORAGE (Phân nhóm XYZ) ===")
print("Phân bố mã XYZCOD:")
print(df_dedicated['XYZCOD'].value_counts())

print("\n=== 4. HYBRID STORAGE (Kết hợp) ===")
print("Phân bố mã XYZCOD:")
print(df_hybrid['XYZCOD'].value_counts())

# So sánh tỷ lệ XYZCOD giữa Dedicated và Hybrid
df_xyz_compare = pd.DataFrame({
    'Dedicated_Count': df_dedicated['XYZCOD'].value_counts(),
    'Hybrid_Count'   : df_hybrid['XYZCOD'].value_counts()
})
df_xyz_compare['Dedicated_%'] = (df_xyz_compare['Dedicated_Count'] / len(df_dedicated) * 100).round(2)
df_xyz_compare['Hybrid_%']    = (df_xyz_compare['Hybrid_Count'] / len(df_hybrid) * 100).round(2)

print("\n=== BẢNG SO SÁNH PHÂN BỔ XYZ GIỮA DEDICATED VÀ HYBRID ===")
display(df_xyz_compare)
```

## 7. Tổng Kết Toàn Diện EDA & Đề Xuất Cho ETL / Data Warehouse

### Bảng tổng hợp 9 Dataset:

| STT | Tên File | Số dòng | Số cột | Dấu phân cách (`sep`) | Lưu ý chất lượng dữ liệu | Vai trò trong Data Warehouse |
|:---:|:---|:---:|:---:|:---:|:---|:---|
| 1 | `Product.csv` | 208 | 3 | `;` | Sạch, 100% unique `Reference`, chia 3 nhóm A/B/C | `DIM_PRODUCT` |
| 2 | `Storage_Location.csv` | 2,292 | 5 | `,` | Tọa độ $X \in [86, 666]$, $Y \in [0, 1440]$, $Z \in [1, 4]$ | `DIM_LOCATION` |
| 3 | `Support_Points_Navigation.csv` | 44 | 2 | `;` | Tọa độ chuỗi `(x, y, z)` cần parse thành float | Hỗ trợ tính routing / cự ly |
| 4 | `Customer_Order.csv` | 122,370 | 9 | `;` (utf-8-sig) | 16 dòng thiếu `Size (US)`, thời gian từ 01/2023 - 10/2023 | `FACT_ORDER` & `DIM_DATE` |
| 5 | `Picking_Wave.csv` | 215,192 | 6 | `;` (utf-8-sig) | 22 vị trí ngoại lệ (vd: `EX-01`), 22 operators | `FACT_PICKING` & `DIM_OPERATOR` |
| 6 | `Random_Storage.csv` | 2,292 | 19 | `,` | 18 slot dạng chuỗi `product;qty`, cần unpivot | `FACT_STORAGE` (Strategy: Random) |
| 7 | `Class_Based_Storage.csv` | 2,341 | 20 | `;` (utf-8-sig) | Có cột `ABCCOD`, 18 slot cần unpivot | `FACT_STORAGE` (Strategy: ClassBased) |
| 8 | `Dedicated_Storage.csv` | 2,340 | 20 | `;` (utf-8-sig) | Có cột `XYZCOD`, 18 slot cần unpivot | `FACT_STORAGE` (Strategy: Dedicated) |
| 9 | `Hybrid_Storage.csv` | 2,340 | 20 | `;` (utf-8-sig) | Có cột `XYZCOD`, 18 slot cần unpivot | `FACT_STORAGE` (Strategy: Hybrid) |

### Các quy tắc kỹ thuật cần thực thi trong Pipeline ETL:
1. **Encoding & Separator**: Dùng `sep=';'` và `encoding='utf-8-sig'` cho 7 file; dùng `sep=','` cho `Storage_Location.csv` và `Random_Storage.csv`.
2. **Missing Values**: Điền giá trị trung vị hoặc giá trị mặc định cho 16 bản ghi thiếu `Size (US)` trong `Customer_Order`.
3. **Location Mapping**: Xử lý 22 mã vị trí đặc biệt (như cửa xuất `EX-01`, khu vực `C-014`..) để không vi phạm khóa ngoại trong `FACT_PICKING`.
4. **Data Normalization (Unpivot/Melt)**: Chuyển 18 cột slot dạng wide trong 4 file chiến lược thành dạng long format chuẩn hóa (`strategy_key, location_key, product_key, slot, quantity`) để nạp vào `FACT_STORAGE`.
